```python
# Env Setup
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
# Data
index = [
    "Original",
    "P/A (25% Confidential)",
    "P/A (50% Confidential)",
    "P/A (75% Confidential)",
    # "P/A (100% Confidential)",
]
f1score = [
    52.37,  # 60.15 with 10 doc
    40.28,
    37.73,  # 40.62, 48.26 using fid
    31.87,  # 27.44, 28.92 using fid
]
emscore = [
    44.62,  # 47.69 with 10 doc
    32.31,
    32.31,  # 35.38, 40.00 using fid
    26.15,  # 24.62, 21,54 using fid
]
time = [
    199.3,
    274.8,
    277.2,
    270,
]
data = pd.DataFrame(
    {
        "Index": index,
        "F1": f1score,
        "EM": emscore,
        # "Time": time,
    }
)
```

```python
# Reshape data for easier comparison
data_melted = data.melt(id_vars="Index", var_name="Metric", value_name="Value")

# Plot with seaborn
sns.set_theme(style="whitegrid", rc={"grid.alpha": 0.3})
plt.figure(figsize=(6, 6), dpi=200)
ax = sns.barplot(x="Metric", y="Value", hue="Index", data=data_melted, palette="muted", width=0.5,  legend=False)

# Add labels and title
ax.set_title("Metrics Comparison", fontsize=12)
ax.set_xlabel("Metric", fontsize=12)
ax.set_ylabel("Score", fontsize=12)
# plt.legend(
#     title="Category",
#     loc="lower center",
#     bbox_to_anchor=(0.5, 1.05),
#     fontsize=12,
#     title_fontsize=14,
# )

# Show plot
plt.tight_layout()
plt.show()
```
