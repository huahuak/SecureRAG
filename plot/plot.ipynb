```python
import numpy as np
import matplotlib.pyplot as plt

a = np.linspace(0, 1, 500)

# 原始函数
f1 = a * (1 - a)
f2 = a**2
f3 = (1 - a) ** 2

# 总和
total = f1 + f2 + f3
# 归一化
f1_norm = f1 / total
f2_norm = f2 / total
f3_norm = f3 / total

# # 放入数组
# values = np.array([f1, f2, f3])
# # 对数归一化
# epsilon = 1e-8
# values_log = np.log(values + epsilon)
# values_norm = (values_log - values_log.min()) / (values_log.max() - values_log.min())
# # 赋值回原来的变量
# f1_norm, f2_norm, f3_norm = values_norm

# 绘图
plt.figure(figsize=(8, 6))
plt.plot(a, f1_norm, label="f1 normalized")
plt.plot(a, f2_norm, label="f2 normalized")
plt.plot(a, f3_norm, label="f3 normalized")

plt.xlabel("a")
plt.ylabel("Normalized Function Value")
plt.title("Three Functions Normalized to Sum 1")
plt.legend()
plt.grid(True)
plt.show()
```
